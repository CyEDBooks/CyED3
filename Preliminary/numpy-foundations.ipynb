```python
import numpy as np
```

## 1. Arrays and their structure

An array has a fixed number of dimensions and one data type. A two-dimensional array can represent a table or matrix. Its shape is `(rows, columns)`.

```python
table = np.array([
    [0, 1, 1],
    [1, 0, 1],
    [1, 1, 0],
])

print(table)
print("shape:", table.shape)
print("dimensions:", table.ndim)
print("elements:", table.size)
print("dtype:", table.dtype)
```

Useful constructors communicate intent: `zeros` initializes an empty numerical table, `ones` initializes a constant table, `arange` creates evenly spaced integer-like values, and `linspace` creates a specified number of evenly spaced values.

```python
print(np.zeros((2, 3), dtype=int))
print(np.ones(4, dtype=int))
print(np.arange(0, 10, 2))
print(np.linspace(0, 1, 5))
```

## 2. Indexing and slicing

Separate dimensions with commas. As in Python strings and lists, a slice includes its start and excludes its stop.

```python
values = np.arange(1, 13).reshape(3, 4)
print(values)
print("row 1:", values[1, :])
print("column 2:", values[:, 2])
print("upper-right block:\n", values[:2, 2:])
```

### Views and copies

Basic slicing normally returns a **view** that shares data with the original array. This is efficient, but modifying the view also modifies the original. Use `.copy()` when you need independent data.

```python
original = np.array([10, 20, 30, 40, 50])
view = original[1:4]
independent = original[1:4].copy()

view[0] = 999
independent[-1] = -1

print("original:", original)
print("view:", view)
print("copy:", independent)
```

## 3. Boolean masks

A comparison produces an array of Boolean values. Use that mask to select exactly the entries satisfying a condition. Combine conditions with `&` and `|`, and parenthesize each comparison.

```python
numbers = np.arange(16)
mask = (numbers % 2 == 0) & (numbers > 5)

print(mask)
print(numbers[mask])
```

## 4. Vectorized operations and axes

Array operations apply to many elements without an explicit Python loop. For a two-dimensional array, `axis=0` reduces down the rows and returns one result per column; `axis=1` reduces across columns and returns one result per row.

```python
scores = np.array([
    [3, 4, 5],
    [5, 5, 4],
    [2, 3, 4],
])

print("twice every score:\n", 2 * scores)
print("column totals:", scores.sum(axis=0))
print("row averages:", scores.mean(axis=1))
```

## 5. Element-wise and matrix multiplication

For arrays, `A * B` multiplies corresponding entries. The operator `A @ B` performs matrix multiplication. They answer different questions.

```python
A = np.array([[1, 1], [0, 1]])
B = np.array([[1, 0], [1, 1]])

print("element-wise:\n", A * B)
print("matrix product:\n", A @ B)
```

### Application: paths in a directed graph

If `adjacency[i, j]` is 1 when there is an edge from node `i` to node `j`, then entry `(i, j)` of `adjacency @ adjacency` counts directed walks of length two from `i` to `j`.

```python
adjacency = np.array([
    [0, 1, 1],
    [0, 0, 1],
    [1, 0, 0],
])
walks_length_2 = adjacency @ adjacency

print(adjacency)
print("walks of length 2:\n", walks_length_2)
print("from node 0 to node 2:", walks_length_2[0, 2])
```
