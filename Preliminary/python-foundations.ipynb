## 2. Strings as words

A string is an ordered, immutable sequence of characters. Indexing selects one position; slicing selects a half-open interval: the start is included and the stop is excluded.

```python
word = "automata"

print(word[0])       # first symbol
print(word[-1])      # final symbol
print(word[:4])      # prefix of length 4
print(word[4:])      # remaining suffix
print(word[::-1])    # reversed word
print(len(word))
```

## 3. Conditions and loops

A condition produces `True` or `False`. A loop repeats a block. In the next example, the variable `valid` is an accumulator: once a forbidden symbol appears, it remains false.

```python
alphabet = {"0", "1"}
word = "010011"
valid = True

for symbol in word:
    if symbol not in alphabet:
        valid = False
        break

print(f"Is {word!r} over {alphabet}? {valid}")
```

## 4. Choosing a collection

- A **list** is ordered, mutable, and may contain duplicates.
- A **tuple** is ordered and immutable.
- A **set** stores unique hashable values and supports set operations.
- A **dictionary** maps unique keys to values.

Choose according to the mathematical role of the data, not by habit.

```python
symbols = list("abacaba")
alphabet = set(symbols)
interval = (0, 3)
counts = {}

for symbol in symbols:
    counts[symbol] = counts.get(symbol, 0) + 1

print(symbols)
print(alphabet)
print(interval)
print(counts)
```

Set operators mirror familiar mathematical operations: `|` is union, `&` is intersection, `-` is difference, and `<=` tests the subset relation.

```python
left = {"q0", "q1", "q2"}
right = {"q1", "q3"}

print("union:", left | right)
print("intersection:", left & right)
print("difference:", left - right)
print({"q0", "q2"} <= left)
```

## 5. Functions and tests

A useful function has a focused responsibility, explicit parameters, and a return value. A docstring states its contract. Small `assert` statements provide executable examples and reveal regressions.

```python
def is_word_over(word, alphabet):
    """Return True exactly when every symbol in word belongs to alphabet."""
    return all(symbol in alphabet for symbol in word)


binary = {"0", "1"}
assert is_word_over("", binary)
assert is_word_over("10110", binary)
assert not is_word_over("10210", binary)
print("All tests passed.")
```

A dictionary whose keys are `(state, symbol)` tuples is a direct representation of a deterministic transition function.

```python
transition = {
    ("even", "0"): "even",
    ("even", "1"): "odd",
    ("odd", "0"): "odd",
    ("odd", "1"): "even",
}

state = "even"
for symbol in "1011":
    state = transition[(state, symbol)]

print("final state:", state)
```
